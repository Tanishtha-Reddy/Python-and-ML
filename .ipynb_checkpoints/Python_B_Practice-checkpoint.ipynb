#### Exercise 1. Arithmetic Product and Conditional Logic

```python
def arithmetic(a,b):
    product = a*b
    sums = a+b
    if product<=1000:
        return product
    else:
        return sums
```

```python
a = float(input())
b = float(input())

result = arithmetic(a,b)
print(result)
```

#### Exercise 10. Finding Extremes (Min/Max) in a List


```python
nums = [1,2,3,4,10,20,50, 90, 1000,8000,100000,0]
largest = nums[0]
smallest = nums[0]
for num in nums[1:]:
    if num < smallest:
        smallest = num
print(smallest)
for num in nums[1:]:
    if num > largest:
        largest = num
print(largest)
    
```

```python
print(min(nums))
print(max(nums))
```

#### Exercise 16. Numerical Palindrome Check

```python
number = int(input())
num = str(number)
if num == num[::-1]:
    print(f"Number {num} is a palindrome number.")
else:
    print(f"Number {num} is not a palindrome number.")
```

#### Exercise 17. Merging Lists with Parity Filtering

```python
list1 = [int(x) for x in input("Enter list 1:").split()]
list2 = [int(x) for x in input("Enter list 2:").split()]
list3 = []
for i in list1:
    if i%2 != 0:
        list3.append(i)
for j in list2:
    if j%2 == 0:
        list3.append(j)
print(list3)
list3_parity = list3.sort()
print(list3)
```

#### Exercise 19. Multi-Tiered Income Tax Calculation

```python
income = 45000
tax = 0

if income <= 10000:
    tax = 0
elif income <= 20000:
    tax = (income-10000)*10/100
else:
    tax = 0+(10000 *10/100)
    tax = tax + (income - 20000)*20/100
print(tax)
```

#### Exercise 21. Downward Half-Pyramid Pattern


```python
rows = int(input())
#range(start, stop, step)
for i in range(rows,0,-1):
    for j in range(0,i):
        print("*", end = " ")
    print("\n")
```

#### Exercise 22. Custom Exponentiation Function


```python
def exponent(base, exp):
    if exp == 0:
        return 1
    result = 1
    for i in range(exp):
        result = base * result
    return result

exponent(5,2)
```

#### Exercise 23. Check Palindrome Number

```python
def palindrome(num):
    numb = str(num)
    print(f"Original number {num}")
    if numb == numb[::-1]:
        print("Yes, given number is a palindrome")
    else:
        print("No, given number is not a palindrome")
palindrome(121)
```
